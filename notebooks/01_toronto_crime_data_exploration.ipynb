```python
df["OCC_DATE"] = pd.to_datetime(df["OCC_DATE"], errors="coerce")

print("Missing or invalid occurrence dates:",
      df["OCC_DATE"].isna().sum())

print("\nEarliest occurrence date:",
      df["OCC_DATE"].min())

print("\nLatest occurrence date:",
      df["OCC_DATE"].max())

print("\nRecords by occurrence year:")
print(df["OCC_DATE"].dt.year.value_counts().sort_index().tail(15))
```

```python
nsa_df = df[df["NEIGHBOURHOOD_158"] == "NSA"]

print("NSA records:", len(nsa_df))

print("\nRecords by year:")
print(nsa_df["OCC_DATE"].dt.year.value_counts().sort_index().tail(15))

print("\nCrime categories:")
print(nsa_df["CSI_CATEGORY"].value_counts())

print("\nHOOD_158 values:")
print(nsa_df["HOOD_158"].value_counts().head(10))
```

```python
print("NSA records:", len(nsa_df))

print("\nMissing coordinates:")
print(nsa_df[["LAT_WGS84", "LONG_WGS84"]].isna().sum())

print("\nCrime categories:")
print(nsa_df["CSI_CATEGORY"].value_counts())

print("\nPolice divisions:")
print(nsa_df["DIVISION"].value_counts())
```

```python
crime_df = df[
    (df["OCC_DATE"] >= "2014-01-01") &
    (df["NEIGHBOURHOOD_158"] != "NSA")
].copy()

print("Modelling dataset shape:", crime_df.shape)

print("Date range:",
      crime_df["OCC_DATE"].min(),
      "to",
      crime_df["OCC_DATE"].max())

print("Unique neighbourhoods:",
      crime_df["NEIGHBOURHOOD_158"].nunique())

print("Missing neighbourhoods:",
      crime_df["NEIGHBOURHOOD_158"].isna().sum())
```

```python
print("Total records:", len(crime_df))

print("Unique event IDs:",
      crime_df["EVENT_UNIQUE_ID"].nunique())

print("Duplicate event IDs:",
      crime_df["EVENT_UNIQUE_ID"].duplicated().sum())
```

```python
duplicate_df = crime_df[
    crime_df["EVENT_UNIQUE_ID"].duplicated(keep=False)
].sort_values("EVENT_UNIQUE_ID")

print("Rows associated with repeated event IDs:",
      len(duplicate_df))

print("\nNumber of event IDs appearing more than once:",
      duplicate_df["EVENT_UNIQUE_ID"].nunique())

print("\nDistribution of records per event ID:")
print(
    duplicate_df.groupby("EVENT_UNIQUE_ID")
    .size()
    .value_counts()
    .sort_index()
)
```

```python
duplicate_df[
    [
        "EVENT_UNIQUE_ID",
        "OCC_DATE",
        "OFFENCE",
        "CSI_CATEGORY",
        "NEIGHBOURHOOD_158"
    ]
].head(15)
```

```python
print("Exact duplicate rows:",
      crime_df.duplicated().sum())

print("Unique event IDs:",
      crime_df["EVENT_UNIQUE_ID"].nunique())

print("Total records:",
      len(crime_df))
```

```python
event_check = crime_df.groupby("EVENT_UNIQUE_ID").agg(
    occurrence_dates=("OCC_DATE", "nunique"),
    neighbourhoods=("NEIGHBOURHOOD_158", "nunique"),
    crime_categories=("CSI_CATEGORY", "nunique")
)

print("Events with multiple occurrence dates:",
      (event_check["occurrence_dates"] > 1).sum())

print("Events with multiple neighbourhoods:",
      (event_check["neighbourhoods"] > 1).sum())

print("Events with multiple crime categories:",
      (event_check["crime_categories"] > 1).sum())
```

```python
monthly_crime = (
    crime_df
    .groupby([
        pd.Grouper(key="OCC_DATE", freq="MS"),
        "NEIGHBOURHOOD_158"
    ])["EVENT_UNIQUE_ID"]
    .nunique()
    .reset_index(name="crime_count")
)

print("Dataset shape:", monthly_crime.shape)

print("\nDate range:")
print(monthly_crime["OCC_DATE"].min(),
      "to",
      monthly_crime["OCC_DATE"].max())

print("\nUnique neighbourhoods:",
      monthly_crime["NEIGHBOURHOOD_158"].nunique())

print("\nFirst five records:")
display(monthly_crime.head())
```

```python
all_months = pd.date_range(
    start="2014-01-01",
    end="2026-06-01",
    freq="MS"
)

all_neighbourhoods = sorted(
    crime_df["NEIGHBOURHOOD_158"].unique()
)

expected_pairs = pd.MultiIndex.from_product(
    [all_months, all_neighbourhoods],
    names=["OCC_DATE", "NEIGHBOURHOOD_158"]
)

observed_pairs = pd.MultiIndex.from_frame(
    monthly_crime[["OCC_DATE", "NEIGHBOURHOOD_158"]]
)

missing_pairs = expected_pairs.difference(observed_pairs)

print("Expected combinations:", len(expected_pairs))
print("Observed combinations:", len(observed_pairs))
print("Missing combinations:", len(missing_pairs))
```

```python
missing_df = pd.DataFrame(
    missing_pairs.tolist(),
    columns=["OCC_DATE", "NEIGHBOURHOOD_158"]
)

print(missing_df.to_string(index=False))
```

```python
for _, row in missing_df.iterrows():
    neighbourhood = row["NEIGHBOURHOOD_158"]
    month = row["OCC_DATE"]

    print(
        month.strftime("%Y-%m"),
        neighbourhood,
        "| Records in previous month:",
        monthly_crime[
            (monthly_crime["NEIGHBOURHOOD_158"] == neighbourhood) &
            (monthly_crime["OCC_DATE"] == month - pd.DateOffset(months=1))
        ]["crime_count"].sum(),
        "| Records in next month:",
        monthly_crime[
            (monthly_crime["NEIGHBOURHOOD_158"] == neighbourhood) &
            (monthly_crime["OCC_DATE"] == month + pd.DateOffset(months=1))
        ]["crime_count"].sum()
    )
```

```python
for _, row in missing_df.iterrows():
    month = row["OCC_DATE"]
    neighbourhood = row["NEIGHBOURHOOD_158"]

    count = crime_df[
        (crime_df["OCC_DATE"].dt.to_period("M") == month.to_period("M")) &
        (crime_df["NEIGHBOURHOOD_158"] == neighbourhood)
    ]["EVENT_UNIQUE_ID"].nunique()

    print(month.strftime("%Y-%m"), neighbourhood, count)
```

```python
complete_monthly_crime = (
    monthly_crime
    .set_index(["OCC_DATE", "NEIGHBOURHOOD_158"])
    .reindex(expected_pairs, fill_value=0)
    .reset_index()
)

print("Dataset shape:", complete_monthly_crime.shape)
print("Total records:", len(complete_monthly_crime))
print("Unique neighbourhoods:",
      complete_monthly_crime["NEIGHBOURHOOD_158"].nunique())
print("Unique months:",
      complete_monthly_crime["OCC_DATE"].nunique())
print("Missing crime counts:",
      complete_monthly_crime["crime_count"].isna().sum())
print("Zero-count combinations:",
      (complete_monthly_crime["crime_count"] == 0).sum())
```

```python
complete_monthly_crime.to_csv(
    "../data/processed/monthly_neighbourhood_crime.csv",
    index=False
)
```
